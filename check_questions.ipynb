```python
df_maven = df_maven[['Model','question_number','Question','Question.Package.Name','mmg_element','cste_element','Message']]
```

```python
def clean_up(name):
    new_name =''
    for ch in name:
        # If space, digits, uppercase letters, lowercase letters, or underscore, add it to new name otherwise ignore
        if (ord(ch)==32 or 48<=ord(ch)<=57 or 65<=ord(ch)<=90 or 97<=ord(ch)<=122 or ord(ch)==95 ):
            new_name+=ch
    return new_name.strip()
```

```python
df_qs.loc[:,'QUESTION_ID'] = [ clean_up(q) for q in df_qs.loc[:,'QUESTION_ID']]
df_maven.loc[:,'Question'] = [clean_up(q) for q in df_maven.loc[:,'Question']]
```

```python
df_merge = pd.merge(df_qs,df_maven,left_on=['QUESTION_ID','Model'],right_on=['Question','Model'],how='outer',indicator=True)
```
