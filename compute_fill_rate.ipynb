```python
# Code to get fillrate for missing elements from MVP
import pandas as pd
```

```python
# Input file of missing elements
df_missing = pd.read_csv('output/fill_rate_missing_elements_ver2_annotated.csv',na_filter=False)
len(df_missing)
```

```python
df_missing
```

```python
# Group by parent_question_id and model_name to get the num of values reported
df_group = df_missing.groupby(by=['parent_question_id','model_name'])['count_of_values'].sum().reset_index(name='count_of_values')
```

```python
df_group
```

```python
df_group.loc[:,'total_cases'] =[536547 if m=='General' else 87739 if m=='Hepatitis' else 210577 if m=='STD' else 33910 if m== 'Lead' else 0 for m in df_group.loc[:,'model_name']]
```

```python
df_group
```

```python
df_group.loc[:,'fill_rate'] = [round(c*100.0/t,3) for c,t in zip(df_group.loc[:,'count_of_values'],df_group.loc[:,'total_cases'])]
```

```python
df_group
```

```python
df_group.to_csv('output/fill_rate_missing_elements_both_parent_child.csv',index=False)
```
