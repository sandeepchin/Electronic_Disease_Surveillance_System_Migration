```python
import pandas as pd
```

```python
df_maven = pd.read_csv('input/questions_by_model.csv',na_filter=False)
len(df_maven)
```

```python
df_maven.info()
```

```python
def join_all_except_last(question,answer):

    # Check if answer matches last part of the question
    start=len(question)-len(answer)
    if(question[start:]==answer):
        return question[:start].strip('_')
    else:
        return question
```

```python
df_maven.loc[:,'parent_question_id']  = [ join_all_except_last(q,a) for q,a in zip(df_maven.loc[:,'QUESTION_ID'],df_maven.loc[:,'value'])]
```

```python
df_maven
```

```python
# Group by parent_question_id and model_name to get the num of values reported in each group
# reset_index() to convert dataframegroupby object to dataframe
df_group = df_maven.groupby(by=['MODEL_NAME','parent_question_id']).count().reset_index()
```

```python
df_group
```

```python
df_group[['MODEL_NAME','parent_question_id']].to_csv('output/model_and_parent_questions.csv',index=False)
```

```python
# grouping again to get counts by model
df_group = df_group.groupby(by=['MODEL_NAME']).count().reset_index()
```
