```python
import pandas as pd
from fuzzywuzzy import fuzz
from fuzzywuzzy import process
```

```python
df_maven = pd.read_csv('maven_disease_vars_ALL_MODELS_2025-08-05.csv',na_filter=False)
len(df_maven)
```

```python
# Goal is to find duplicates between the three data sources - Maven, MMG and CSTE
df_maven_only = df_maven[df_maven['Source']=='Maven']
len(df_maven_only)
```

```python
df_maven_only
```

```python
df_mmg = df_maven[df_maven['Source']=='MMG']
len(df_mmg)
```

```python
df_mmg
```

```python
# Compare the data elements in df_maven_only and df_mmg

# Using fuzz.token_set_ratio() - ignores duplicate words and order
df_mmg.loc[:,'score'] = [fuzz.token_set_ratio(mmg,maven) for (mmg,maven) in zip(df_mmg.loc[:,'MMG Data Element Name'],df_mmg.loc[:,'Question - Maven Data Element'])]
```

```python
df_mmg[['Question - Maven Data Element','MMG Data Element Name','score']].to_csv('compare_maven_mmg.csv',index=False)
```

```python
df_mmg
```

```python
df_cste = df_maven[df_maven['Source']=='CSTE']
len(df_cste)
```

```python
df_cste
```

```python
# Compare the data elements in df_maven_only and df_mmg

# Using fuzz.token_set_ratio() - ignores duplicate words and order
df_cste.loc[:,'score'] = [fuzz.token_set_ratio(cste,maven) for (cste,maven) in zip(df_cste.loc[:,'CSTE Data Element'],df_cste.loc[:,'Question - Maven Data Element'])]
```

```python
df_cste[['Question - Maven Data Element','CSTE Data Element','score']].to_csv('compare_maven_cste.csv',index=False)
```
