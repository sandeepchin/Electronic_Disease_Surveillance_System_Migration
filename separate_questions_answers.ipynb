```python
# Goal is to identify unique questions by removing the answers appended to end of questions
questions = df_qa.loc[:,'QUESTION_ID']
answers = df_qa.loc[:,'value']
for i in range(len(questions)):
    question = questions[i].strip()
    tokens = question.split('_')
    #Compare last token with value
    value = answers[i].strip()
    if tokens[-1]==value:
        question = question.strip(value).strip('_')
    questions[i] = question

df_qa.loc[:,'QUESTION_ID'] = questions
```
